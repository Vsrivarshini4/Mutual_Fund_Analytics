```python
nav_url = "https://raw.githubusercontent.com/Vsrivarshini4/Mutual_Fund_Analytics/main/data/raw/02_nav_history.csv"

nav_history = pd.read_csv(nav_url)

nav_history.head()
```

```python
nav_history["date"] = pd.to_datetime(nav_history["date"])

daily_avg_nav = nav_history.groupby("date")["nav"].mean()

daily_avg_nav.plot(figsize=(12,5))

plt.title("Average NAV Trend Over Time")
plt.xlabel("Date")
plt.ylabel("Average NAV")
plt.tight_layout()
plt.show()
```

```python
fund_nav = nav_history.merge(
    fund_master[["amfi_code", "scheme_name"]],
    on="amfi_code",
    how="left"
)

fund_nav.head()
```

```python
latest_nav = fund_nav.sort_values("date").groupby("scheme_name").tail(1)

latest_nav[["scheme_name", "date", "nav"]]
```

```python
latest_nav_sorted = latest_nav.sort_values("nav", ascending=False)

latest_nav_sorted.plot(
    x="scheme_name",
    y="nav",
    kind="bar",
    figsize=(12, 6),
    legend=False
)

plt.title("Latest NAV by Mutual Fund Scheme")
plt.xlabel("Fund Scheme")
plt.ylabel("Latest NAV")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
first_nav = fund_nav.sort_values("date").groupby("scheme_name").first()
last_nav = fund_nav.sort_values("date").groupby("scheme_name").last()

nav_change = pd.DataFrame({
    "First NAV": first_nav["nav"],
    "Latest NAV": last_nav["nav"]
})

nav_change["Change %"] = (
    (nav_change["Latest NAV"] - nav_change["First NAV"])
    / nav_change["First NAV"]
) * 100

nav_change.sort_values("Change %", ascending=False).head(10)
```

```python
nav_change.sort_values("Change %", ascending=False).head(10)
```

```python
top_10_nav = nav_change.sort_values("Change %", ascending=False).head(10)

top_10_nav["Change %"].sort_values().plot(
    kind="barh",
    figsize=(10, 6)
)

plt.title("Top 10 Mutual Fund Schemes by NAV Change")
plt.xlabel("NAV Change (%)")
plt.ylabel("Fund Scheme")
plt.tight_layout()
plt.show()
```

```python
import sqlite3

conn = sqlite3.connect("mutual_fund.db")

fund_master.to_sql(
    "fund_master",
    conn,
    if_exists="replace",
    index=False
)

print("Fund Master loaded into SQL successfully!")
```

```python
query = """
SELECT category, COUNT(*) AS scheme_count
FROM fund_master
GROUP BY category;
"""

sql_result = pd.read_sql_query(query, conn)

sql_result
```

```python
query = """
SELECT risk_category, COUNT(*) AS scheme_count
FROM fund_master
GROUP BY risk_category
ORDER BY scheme_count DESC;
"""

sql_result = pd.read_sql_query(query, conn)

sql_result
```

```python
query = """
SELECT
    risk_category,
    ROUND(AVG(expense_ratio_pct), 3) AS avg_expense_ratio
FROM fund_master
GROUP BY risk_category
ORDER BY avg_expense_ratio DESC;
"""

sql_result = pd.read_sql_query(query, conn)

sql_result
```

```python
query = """
SELECT
    fund_house,
    COUNT(*) AS scheme_count
FROM fund_master
GROUP BY fund_house
ORDER BY scheme_count DESC;
"""

sql_result = pd.read_sql_query(query, conn)

sql_result
```

```python
query = """
SELECT
    fund_house,
    ROUND(AVG(expense_ratio_pct), 3) AS avg_expense_ratio
FROM fund_master
GROUP BY fund_house
ORDER BY avg_expense_ratio DESC;
"""

sql_result = pd.read_sql_query(query, conn)

sql_result
```
